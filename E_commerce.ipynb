```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
url = (
    "https://raw.githubusercontent.com/araj2/customer-database/refs/heads/master/Ecommerce%20Customers.csv"
)

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
#Data Cleaning
print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nStatistical summary:")
display(df.describe())

# Remove duplicate rows if any
df = df.drop_duplicates()

print("\nDataset shape after removing duplicates:", df.shape)
```

```python

print("Average yearly spending:",
      round(df["Yearly Amount Spent"].mean(), 2))

print("Highest yearly spending:",
      round(df["Yearly Amount Spent"].max(), 2))

print("Lowest yearly spending:",
      round(df["Yearly Amount Spent"].min(), 2))

plt.figure(figsize=(8, 5))
sns.histplot(df["Yearly Amount Spent"], kde=False)

plt.title("Distribution of Yearly Customer Spending")
plt.xlabel("Yearly Amount Spent")
plt.ylabel("Number of Customers")
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Time on Website",
    y="Yearly Amount Spent"
)

plt.title("Website Time vs Yearly Spending")
plt.xlabel("Time on Website")
plt.ylabel("Yearly Amount Spent")
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Time on App",
    y="Yearly Amount Spent"
)

plt.title("App Time vs Yearly Spending")
plt.xlabel("Time on App")
plt.ylabel("Yearly Amount Spent")
plt.show()
```

```python

print("E-COMMERCE CUSTOMER ANALYSIS SUMMARY")
print("-" * 40)

print("Total customer records:", len(df))
print("Average yearly spending:",
      round(df["Yearly Amount Spent"].mean(), 2))

print("\nAverage time on app:",
      round(df["Time on App"].mean(), 2))

print("Average time on website:",
      round(df["Time on Website"].mean(), 2))

print("\nTop 5 customers by yearly spending:")

display(
    df.nlargest(5, "Yearly Amount Spent")[
        ["Email", "Yearly Amount Spent"]
    ]
)
```
